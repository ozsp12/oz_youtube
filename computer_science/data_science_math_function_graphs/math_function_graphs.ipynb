# Functions

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, a=2.5, b=0.5, c=1.5, d=2.5): return (x + a) * (x + b) * (x - c) * (x - d)

def create_polynomial_plot(f, *args, fig_size=(4, 4), x_min=-5, x_max=5, n_points=400, **kwargs):
    x = np.linspace(x_min, x_max, n_points)
    y = f(x, *args, **kwargs)
    fig, ax = plt.subplots(figsize=fig_size)
    fig.patch.set_facecolor('#a9c9e6')
    ax.set_facecolor('#a9c9e6')
    ax.plot(x, y, 'k', linewidth=2)
    ax.axhline(0, color='k', linewidth=1)
    ax.axvline(0, color='k', linewidth=1)
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(-20, 20)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_title(r'Continuidade em todos os valores $x$', fontsize=12, pad=15)
    plt.tight_layout()
    plt.close(fig)
    return fig


def plot_jump_discontinuity(xmin_left=-5.0, xmax_right=5.0, a=0.0, fig_size=(4, 4), n_points=400):
    x_left  = np.linspace(xmin_left, a, n_points, endpoint=False)
    y_left  = 0.5 * (x_left + 1.5)**2 - 2
    x_right = np.linspace(a, xmax_right, n_points, endpoint=False)
    y_right = -0.3 * (x_right - 0.5)**2 + 1.5
    yL = 0.5 * (a + 1.5)**2 - 2
    yR = -0.3 * (a - 0.5)**2 + 1.5

    fig, ax = plt.subplots(figsize=fig_size)
    ax.plot(x_left, y_left, 'k', linewidth=2)
    ax.plot(x_right, y_right, 'k', linewidth=2)
    ax.plot(a, yL, 'ko', mfc='white', markersize=8, clip_on=False)
    ax.plot(a, yR, 'ko', mfc='black', markersize=8, clip_on=False)
    ax.axhline(0, color='k', linewidth=1)
    ax.axvline(0, color='k', linewidth=1)
    ax.set_title('Descontinuidade de pulo (ou salto)', pad=12)

    xmin = min(x_left.min(), a)
    xmax = max(x_right.max(), a)
    ymin = min(y_left.min(), y_right.min(), yL, yR)
    ymax = max(y_left.max(), y_right.max(), yL, yR)
    padx = 0.25 * (xmax - xmin)
    pady = 0.25 * (ymax - ymin)
    ax.set_xlim(xmin - padx, xmax + padx)
    ax.set_ylim(ymin - pady, ymax + pady)
    ax.set_xticks([]); ax.set_yticks([])
    plt.subplots_adjust(left=0.12, right=0.95, bottom=0.12, top=0.90)
    plt.close(fig)
    return fig
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Configuração geral ---
x = np.linspace(-3, 3, 400)
y = x**3 - 3*x  # Função base só para ter forma semelhante à da imagem

fig, axes = plt.subplots(2, 3, figsize=(12, 7))
axes = axes.flatten()

# 1. Continuidade em todos os valores x
axes[0].plot(x, y, 'k')
axes[0].set_title('Continuidade em todos os valores x')

# 2. Descontinuidade removível (ponto não definido em a)
a = 0
axes[1].plot(x, y, 'k')
axes[1].plot(a, np.nan, 'ko', mfc='white')  # ponto vazio
axes[1].set_title('Descontinuidade removível')

# 3. Descontinuidade removível (ponto deslocado)
axes[2].plot(x, y, 'k')
axes[2].plot(a, -2, 'ko', mfc='white')  # buraco
axes[2].plot(a, 2, 'ko', mfc='black')   # ponto em outro valor
axes[2].text(a + 0.2, 2, r'$f(a)$', fontsize=10)
axes[2].set_title('Descontinuidade removível')

# 4. Descontinuidade de pulo (ou salto)
x1 = np.linspace(-3, 0, 200)
x2 = np.linspace(0, 3, 200)
axes[3].plot(x1, x1**2 - 1, 'k')
axes[3].plot(x2, -x2**2 + 2, 'k')
axes[3].plot(0, (-1), 'ko', mfc='white')  # limite esquerdo
axes[3].plot(0, (2), 'ko', mfc='black')   # limite direito
axes[3].set_title('Descontinuidade de pulo (ou salto)')

# 5. Descontinuidade infinita
x3 = np.linspace(-2, -0.1, 200)
x4 = np.linspace(0.1, 2, 200)
axes[4].plot(x3, 1/x3, 'k')
axes[4].plot(x4, 1/x4, 'k')
axes[4].axvline(0, color='k', linestyle='--', linewidth=1)
axes[4].set_xlim(-2, 2)
axes[4].set_ylim(-5, 5)
axes[4].set_title('Descontinuidade infinita')

# --- Ajustes visuais ---
for ax in axes:
    ax.axhline(0, color='k', linewidth=0.8)
    ax.axvline(0, color='k', linewidth=0.8)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)

# Oculta o sexto subplot vazio
axes[-1].axis('off')

plt.tight_layout()
plt.show()
```
