# Functions

```python
import numpy as np
import matplotlib.pyplot as plt

def chart_histogram(n=1000, mean=50, std=10, seed=42):
    np.random.seed(seed)
    data = np.random.normal(mean, std, n)

    plt.figure(figsize=(8, 5))
    plt.hist(data, bins=30, color="steelblue", edgecolor="black", alpha=0.75)
    plt.title("Histogram of Synthetic Data", fontsize=14)
    plt.xlabel("Value")
    plt.ylabel("Frequency")
    plt.grid(alpha=0.3)
    plt.show()
    return data

def generate_bar_graph():
    categories = ['A', 'B', 'C', 'D', 'E']
    values = [5, 10, 20, 35, 50]

    plt.figure(figsize=(8, 8))
    bars = plt.bar(categories, values, color='steelblue', edgecolor='black', alpha=0.8)

    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height + 0.8, f'{height}',
                 ha='center', va='bottom', fontsize=10)

    plt.title("Bar Graph of Synthetic Data")
    plt.xlabel("Category")
    plt.ylabel("Value")
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()
    
def generate_horizontal_bar_graph():
    categories = ['A', 'B', 'C', 'D', 'E']
    values = [48, 43, 32, 17, 12]

    plt.figure(figsize=(7, 5))
    bars = plt.barh(categories, values, color='seagreen', edgecolor='black', alpha=0.8)

    for bar in bars:
        width = bar.get_width()
        plt.text(width + 0.8, bar.get_y() + bar.get_height()/2, f'{width}',
                 va='center', ha='left', fontsize=10)

    plt.title("Horizontal Bar Graph of Synthetic Data")
    plt.xlabel("Value")
    plt.ylabel("Category")
    plt.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()
    
    
def generate_stacked_bar_chart():
    categories = ['Q1', 'Q2', 'Q3', 'Q4']
    product_A = np.random.randint(20, 60, len(categories))
    product_B = np.random.randint(15, 50, len(categories))
    product_C = np.random.randint(10, 40, len(categories))

    plt.figure(figsize=(8, 8))
    bars_A = plt.bar(categories, product_A, color='steelblue', label='Product A')
    bars_B = plt.bar(categories, product_B, bottom=product_A, color='seagreen', label='Product B')
    bars_C = plt.bar(categories, product_C, bottom=product_A + product_B, color='indianred', label='Product C')

    # Add value labels for each product
    for i in range(len(categories)):
        plt.text(i, product_A[i]/2, f'{product_A[i]}', ha='center', va='center', color='white', fontsize=9)
        plt.text(i, product_A[i] + product_B[i]/2, f'{product_B[i]}', ha='center', va='center', color='white', fontsize=9)
        plt.text(i, product_A[i] + product_B[i] + product_C[i]/2, f'{product_C[i]}', ha='center', va='center', color='white', fontsize=9)

    # Add total labels on top
    totals = product_A + product_B + product_C
    for i, total in enumerate(totals):
        plt.text(i, total + 2, f'{total}', ha='center', va='bottom', fontsize=10)

    plt.title("Stacked Bar Chart of Synthetic Data")
    plt.xlabel("Quarter")
    plt.ylabel("Sales")
    plt.legend()
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()
    
    
def generate_stacked_bar_chart_100():
    categories = ['Q1', 'Q2', 'Q3', 'Q4']
    product_A = np.random.randint(20, 60, len(categories))
    product_B = np.random.randint(15, 50, len(categories))
    product_C = np.random.randint(10, 40, len(categories))

    totals = product_A + product_B + product_C
    product_A_perc = product_A / totals * 100
    product_B_perc = product_B / totals * 100
    product_C_perc = product_C / totals * 100

    plt.figure(figsize=(8, 8))
    plt.bar(categories, product_A_perc, color='steelblue', label='Product A')
    plt.bar(categories, product_B_perc, bottom=product_A_perc, color='seagreen', label='Product B')
    plt.bar(categories, product_C_perc, bottom=product_A_perc + product_B_perc, color='indianred', label='Product C')

    # Add percentage labels
    for i in range(len(categories)):
        plt.text(i, product_A_perc[i]/2, f'{product_A_perc[i]:.0f}%', ha='center', va='center', color='white', fontsize=9)
        plt.text(i, product_A_perc[i] + product_B_perc[i]/2, f'{product_B_perc[i]:.0f}%', ha='center', va='center', color='white', fontsize=9)
        plt.text(i, 100 - product_C_perc[i]/2, f'{product_C_perc[i]:.0f}%', ha='center', va='center', color='white', fontsize=9)

    plt.title("100% Stacked Bar Chart of Synthetic Data")
    plt.xlabel("Quarter")
    plt.ylabel("Percentage (%)")
    plt.ylim(0, 100)
    plt.legend()
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()
```

# Pipeline

## Histogram

```python
data = chart_histogram(2000, mean=100, std=20)
```

## Bar chart

```python
generate_bar_graph()
```

## Horizontal bar chart

```python
generate_horizontal_bar_graph()
```

## Stacked bar chart

```python
generate_stacked_bar_chart()
```

## Stacked bar chart 100%

```python
generate_stacked_bar_chart_100()
```
