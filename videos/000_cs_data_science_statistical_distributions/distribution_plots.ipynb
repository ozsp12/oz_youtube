# Funções

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def creates_dist_synth(n_sample):
    # Set a random seed for reproducibility
    np.random.seed(0)
    # Number of data points
    num_samples = n_sample
    # Define distributions
    distributions = [
        ('Uniform', np.random.uniform(0, 1, num_samples)),
        ('Normal', np.random.normal(0, 1, num_samples)),
        ('Exponential', np.random.exponential(1, num_samples)),
        ('Poisson', np.random.poisson(3, num_samples)),
        ('Binomial', np.random.binomial(10, 0.3, num_samples)),
        ('Geometric', np.random.geometric(0.2, num_samples)),
        ('Logistic', np.random.logistic(0, 1, num_samples)),
        ('Chi-Square', np.random.chisquare(5, num_samples)),
        ('Beta', np.random.beta(2, 5, num_samples)),
        ('Gamma', np.random.gamma(2, 1, num_samples)),
        ('Cauchy', np.random.standard_cauchy(num_samples)),
        ('Laplace', np.random.laplace(0, 1, num_samples)),
        ('Weibull', np.random.weibull(2, num_samples)),
        ('Rayleigh', np.random.rayleigh(1, num_samples)),
        ('Pareto', np.random.pareto(2, num_samples)),
        ('F', np.random.f(2, 5, num_samples)),
        ('Exponential Power', np.random.exponential(2, num_samples)),
        ('Hypergeometric', np.random.hypergeometric(10, 5, 7, num_samples)),
        ('Wald', np.random.wald(2, 3, num_samples)),
        ('Zipf', np.random.zipf(2, num_samples)),
        #('Triangular', np.random.triangular(0, 1, 0.5, num_samples)),
    ]
    # Create a DataFrame with the distributions
    data = {dist[0]: dist[1] for dist in distributions}
    df = pd.DataFrame(data)
    #
    return df


def plot_graph_distr(df, n_cols):
    # Plotting the distributions
    num_graphs = df.shape[1]
    num_cols = n_cols
    # Adjust the number of rows per your preference
    num_rows = int(np.ceil(num_graphs / num_cols))
    # Create subplots based on calculated rows and columns
    fig, axes = plt.subplots(num_rows, num_cols, figsize = (18, 5 * num_rows))
    fig.suptitle('Various Probability Distributions', fontsize=16)
    #
    for i, col in enumerate(df.columns, 1):
        plt.subplot(num_rows, num_cols, i)
        plt.hist(df[col], bins=15, alpha=0.7)
        plt.title(col)
        plt.xlabel('Value')
        plt.ylabel('Frequency')
    # Remove any unused subplots
    for j in range(i, num_rows * num_cols):
        fig.delaxes(axes.flatten()[j])

    #
    plt.tight_layout()
    plt.subplots_adjust(top = 0.95) # Increased top to reduce space
    #plt.show()
    return fig


def plot_graph_distr_obsolete(df, n_cols):
    # Plotting the distributions
    fig, axes = plt.subplots(5, 5, figsize = (15, 20))
    fig.suptitle('Various Probability Distributions', fontsize=16)
    #
    num_graphs = df.shape[1]
    #
    num_cols = n_cols
    # Adjust the number of rows per your preference
    num_rows = int(np.ceil(num_graphs / num_cols))  
    #
    for i, col in enumerate(df.columns, 1):
        plt.subplot(num_rows, num_cols, i)
        plt.hist(df[col], bins=15, alpha=0.7)
        plt.title(col)
        plt.xlabel('Value')
        plt.ylabel('Frequency')
    #
    plt.tight_layout()
    plt.subplots_adjust(top = 0.9)
    #plt.show()
    return fig
```

# Pipeline

```python
df = creates_dist_synth(n_sample = 1000)
fig = plot_graph_distr(df, n_cols = 4)
```
