# Introdução ao Problema — Questão 160 do ENEM 2025

Este estudo analisa a **Questão nº 160 do ENEM 2025**, que envolve a modelagem matemática do enchimento de um recipiente com vazão constante. A distância $D$, em centímetros, da lâmina de água até o tampo da mesa varia conforme o tempo $T$, em minutos, e é descrita por uma função do tipo:

$$
D(T) = k + \tan\big(p(T + m)\big),
$$

onde $k$, $p$ e $m$ são parâmetros reais desconhecidos que devem ser determinados a partir da análise do gráfico fornecido.

O gráfico apresenta:

- Duas assíntotas verticais provenientes da estrutura da função tangente;  
- O valor pontual $D=30$ no instante $T=2{,}5$;  
- O domínio explícito da função no intervalo $0 < T \le 4$;  
- O comportamento crescente característico da tangente, reescalada e deslocada.

A seguir, está a imagem oficial da questão usada como referência para determinar os parâmetros da função:

![Figura da questão](fig01.png)

A partir dessa representação gráfica, o objetivo é construir a expressão algébrica exata de $D(T)$, identificando corretamente os valores de $k$, $p$ e $m$.

# Conceitos teóricos da função tangente aplicada ao problema

A questão do ENEM 2025 apresentada abaixo utiliza uma modelagem baseada em funções do tipo tangente para descrever o crescimento da distância $D$ da lâmina de água ao tampo da mesa em função do tempo $T$. A expressão dada no enunciado é

$$
D(T) = k + \tan\!\big(p\,(T+m)\big),
$$

onde os parâmetros $k$, $p$ e $m$ devem ser determinados pela análise do gráfico. A escolha da função tangente não é arbitrária: ela apresenta assíntotas verticais, crescimento acelerado e comportamento simétrico em torno do seu ponto médio, o que é compatível com recipientes que possuem regiões estreitas seguidas de alargamentos. Para compreender como determinar os parâmetros a partir do gráfico, é fundamental analisar a estrutura teórica da função tangente na forma geral

$$
f(x) = a + b \tan(cx + d).
$$

A tangente possui assíntotas verticais nos pontos em que o argumento satisfaz

$$
cx + d = \frac{\pi}{2} + k\pi, \quad k \in \mathbb{Z}.
\tag{01}
$$

Se considerarmos um ponto $x + \Delta x$ correspondente à próxima assíntota, então

$$
c(x + \Delta x) + d = \frac{\pi}{2} + (k+1)\pi.
\tag{02}
$$

Subtraindo (01) de (02), obtemos

$$
c\,\Delta x = \pi
\quad\Longrightarrow\quad
c = \frac{\pi}{\Delta x}.
$$

Essa relação mostra que **o parâmetro $c$ é determinado diretamente pela distância entre duas assíntotas consecutivas**. 

---

## Interpretação dos parâmetros da função geral

A decomposição da função

$$
f(x) = a + b \tan(cx + d)
$$

permite interpretar cada parâmetro com significado geométrico claro:

- **$a$**: translação vertical do gráfico.  
  Desloca toda a curva para cima ou para baixo.

- **$b$**: amplitude vertical da tangente.  
  Modifica a intensidade das variações; afeta a inclinação e o crescimento da curva.

- **$c$**: mudança de escala horizontal, responsável pelo período da função.  
  O período é dado por  
  $$
  \Delta x = \frac{\pi}{|c|}.
  $$

- **$d$**: ângulo de fase, responsável pelo deslocamento horizontal da curva.  
  A translação horizontal ocorre em  
  $$
  x = -\frac{d}{c}.
  $$

---

## Domínio e imagem da função tangente transformada

A tangente é definida para todos os reais exceto onde ocorre divisão por zero no denominador $\cos(cx + d)$. Assim:

- **Domínio:**  
  $$
  D = \{x \in \mathbb{R} \;|\; x \neq \frac{\pi}{2c} - \frac{d}{c} + k\frac{\pi}{c},\; k\in\mathbb{Z} \}.
  $$

- **Imagem:**  
  $$
  I = \mathbb{R}.
  $$

Independentemente das transformações aplicadas, a função tangente sempre assume todos os valores reais.

---

## Algoritmo de resolução

Três parâmetros a serem determinados
* k: deslocamento vertical
* p: escala horizontal
* m: deslocamento horizontal

O período é determinado como a distância entre as assíntotas consecutivas

$$
\Delta T = \frac{5+2\pi}{2} - \frac{5-2\pi}{2} = 2\pi
$$

O período (\Delta T) da função tangente satisfaz

$$
\Delta T = \frac{\pi}{|p|}
$$

Portanto,

$$
2\pi = \frac{\pi}{|p|} \Rightarrow p = \frac{1}{2}
$$

O deslocamento horizontal é o negativo do ponto médio das assíntotas

$$
m = -\left( \frac{5+2\pi}{2} + \frac{5-2\pi}{2} \right) = -\frac{5}{2}
$$

Para determinar o valor de (k), usa-se o ponto $(2,5,30)$:

$$
30 = k + \tan\left[\frac{1}{2}(2{,}5 - \frac{5}{2})\right]
$$

Como o argumento é zero, então o valor é de $k = 30$, e dessa forma a equação da função descrita no problema do Enem é dada por:

$$
D(T) = 30 + \tan\left[\frac{1}{2}\left(T - \frac{5}{2}\right)\right]
$$

# Visualização da função tangente parametrizada

Este código tem como objetivo **comparar o efeito dos parâmetros** da função 

$$
D(T) = k + \tan\big(p(T+m)\big),
$$ 

por meio de um painel com seis gráficos. Cada subplot utiliza um conjunto diferente de valores \((k,p,m)\), permitindo observar como cada parâmetro altera a forma da curva.

- **k** controla o deslocamento vertical da função.  
- **p** ajusta a escala horizontal, modificando a distância entre assíntotas.  
- **m** define o deslocamento horizontal.  

O código gera um painel 3×2, calcula a função em um domínio contínuo e mascara valores próximos às assíntotas para evitar divergências visuais. O resultado é uma ferramenta clara para estudar transformações da função tangente.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Função geral: D(T) = k + tan(p(T + m))
def D(T, k, p, m):
    return k + np.tan(p * (T + m))

# 2. Parâmetros usados no painel
def gerar_parametros():
    return [
        (30, 0.5, -2.5),   # caso 1 – base
        (20, 0.5, -2.5),   # muda k
        (30, 4.0, -2.5),   # muda p
        (30, 0.5, -1.0),   # muda m
        (10, 0.3, 0.0),    # livre
        (40, 0.8, -3.0)    # livre
    ]

# 3. Domínio de T
def gerar_dominio(T_min=-5, T_max=10, n=5000):
    return np.linspace(T_min, T_max, n)

# 4. Aplicar máscara para esconder divergências
def mascarar_assintotas(y, limite=100):
    return np.ma.masked_where(np.abs(y) > limite, y)

# 5. Plotar uma única curva no eixo fornecido
def plotar_curva(ax, T, k, p, m):
    D_vals = D(T, k, p, m)
    D_vals = mascarar_assintotas(D_vals)

    ax.plot(T, D_vals, color='darkred')
    
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)

    ax.set_title(f"k={k}, p={p}, m={m}")
    ax.set_xlim([-5, 10])
    ax.set_ylim([-50, 50])
    ax.grid(True)

# 6. Criar painel 2 × 3 com todos os parâmetros
def plotar_painel(parametros, T):
    fig, axes = plt.subplots(2, 3, figsize=(15, 8))
    axes = axes.flatten()

    for ax, (k, p, m) in zip(axes, parametros):
        plotar_curva(ax, T, k, p, m)

    plt.tight_layout()
    plt.show()

# 7. Função principal
def main():
    parametros = gerar_parametros()
    T = gerar_dominio()
    plotar_painel(parametros, T)

main()
```

# Gráfico padrão da função tangente

Este código produz gráfico da função tangente padrão. Primeiro, define-se o domínio e calcula-se $y=\tan(x)$. Para evitar explosões visuais próximas às assíntotas, aplica-se uma máscara que oculta valores muito grandes. Em seguida, o código plota a curva suavizada e adiciona as assíntotas verticais nos pontos $x=\pi/2 + k\pi$, destacando o comportamento periódico da função.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Geração do domínio
def gerar_dominio(x_min=-2*np.pi, x_max=2*np.pi, n=5000):
    return np.linspace(x_min, x_max, n)

# 2. Função tangente padrão
def calcular_tangente(x):
    return np.tan(x)

# 3. Aplicar máscara para evitar explosões nas assíntotas
def mascarar_assintotas(y, limite=20):
    return np.ma.masked_where(np.abs(y) > limite, y)

# 4. Adicionar assíntotas (π/2 + kπ)
def plotar_assintotas(ax, k_min=-3, k_max=3):
    for k in range(k_min, k_max + 1):
        xa = np.pi/2 + k*np.pi
        ax.axvline(xa, color='gray', linestyle='--', linewidth=1)

# 5. Configurar eixos, rótulos e estilo
def configurar_estilo(ax):
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)

    ax.set_title("Função Tangente Padrão: $y = \\tan(x)$", fontsize=14)
    ax.set_xlabel("x (radianos)", fontsize=12)
    ax.set_ylabel("y", fontsize=12)

    ax.set_ylim(-10, 10)
    ax.set_xlim(-2*np.pi, 2*np.pi)

    ax.grid(True, linestyle=':', alpha=0.7)

    ax.set_xticks(
        [-2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2, 0,
         np.pi/2, np.pi, 3*np.pi/2, 2*np.pi]
    )
    ax.set_xticklabels(
        [r"$-2\pi$", r"$-\frac{3\pi}{2}$", r"$-\pi$", r"$-\frac{\pi}{2}$", "0",
         r"$\frac{\pi}{2}$", r"$\pi$", r"$\frac{3\pi}{2}$", r"$2\pi$"]
    )

# 6. Plotar tangente no eixo fornecido
def plotar_tangente(ax, x, y_mask):
    ax.plot(x, y_mask, color='darkred', linewidth=2, label='tan(x)')
    ax.legend()

# 7. Função principal para exibir tudo
def plotar_grafico_tangente():
    x = gerar_dominio()
    y = calcular_tangente(x)
    y_mask = mascarar_assintotas(y)

    fig, ax = plt.subplots(figsize=(10, 6))

    plotar_tangente(ax, x, y_mask)
    plotar_assintotas(ax)
    configurar_estilo(ax)

    plt.show()

plotar_grafico_tangente()
```

Este código ilustra como os parâmetros da função geral 

$$f(x) = a + b\tan(cx+d)$$ 

afetam a forma do gráfico. Para isso, define-se um conjunto de seis combinações distintas de parâmetros ((a,b,c,d)), variando separadamente o deslocamento vertical, a amplitude, a escala horizontal e o deslocamento horizontal. O código organiza os gráficos em um painel de 3 colunas por 2 linhas, onde cada subplot exibe a função correspondente ao conjunto de parâmetros especificado. Valores excessivamente grandes, decorrentes das assíntotas da tangente, são mascarados para evitar distorções no gráfico. Cada subplot inclui eixos centrais, grade e título com os parâmetros utilizados, facilitando a comparação direta entre os efeitos de cada transformação. 

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Função geral f(x) = a + b * tan(c*x + d)
def f(x, a, b, c, d):
    return a + b * np.tan(c * x + d)

# 2. Retorna os parâmetros usados no painel
def gerar_parametros():
    return [
        (0, 1, 1, 0),       # tangente padrão
        (2, 1, 1, 0),       # muda a
        (0, 4, 1, 0),       # muda b (escala)
        (0, 1, 0.5, 0),     # muda c (período)
        (0, 1, 1, -1),      # muda d (fase)
        (3, 0.7, 1.5, 0.5)  # variação geral
    ]

# 3. Função para gerar domínio de x
def gerar_dominio(x_min=-10, x_max=10, n=5000):
    return np.linspace(x_min, x_max, n)

# 4. Plota uma única função em um eixo
def plotar_funcao(ax, x, a, b, c, d):
    y = f(x, a, b, c, d)

    # Máscara para evitar explosões próximas às assíntotas
    y = np.ma.masked_where(np.abs(y) > 100, y)

    ax.plot(x, y, color='darkred')

    # Eixos
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)

    # Título
    ax.set_title(f"a={a}, b={b}, c={c}, d={d}")

    # Limites
    ax.set_xlim([-10, 10])
    ax.set_ylim([-50, 50])

    ax.grid(True)

# 5. Monta o painel completo de 2 × 3
def plotar_painel(parametros, x):
    fig, axes = plt.subplots(2, 3, figsize=(15, 8))
    axes = axes.flatten()

    for ax, (a, b, c, d) in zip(axes, parametros):
        plotar_funcao(ax, x, a, b, c, d)

    plt.tight_layout()
    plt.show()

# 6. Função principal
def main():
    parametros = gerar_parametros()
    x = gerar_dominio()
    plotar_painel(parametros, x)

main()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML

# 1. Função geral f(x) = a + b tan(c x + d)
def tangent_function(x, a, b, c, d):
    return a + b * np.tan(c * x + d)

# 2. Gera os dados de x e y
def compute_data(a, b, c, d, x_min, x_max, n_points):
    x = np.linspace(x_min, x_max, n_points)
    y = tangent_function(x, a, b, c, d)
    return x, y

# 3. Configuração da figura e assíntotas
def config_axes(x_min, x_max, y_min, y_max, asymptotes):
    fig, ax = plt.subplots(figsize=(8, 5))

    ax.set_xlim(x_min - 0.05, x_max + 0.05)
    ax.set_ylim(y_min, y_max)

    asymp_lines = []
    for x_asymp in asymptotes:
        line = ax.axvline(x_asymp, color='gray', linestyle='--', linewidth=1)
        asymp_lines.append(line)

    ax.set_title("Animação da Função Geral f(x) = a + b tan(c x + d)")
    ax.set_xlabel("x")
    ax.set_ylabel("f(x)")

    return fig, ax, asymp_lines

# 4. Inicialização da animação
def init_animation(line, point):
    line.set_data([], [])
    point.set_data([], [])
    return line, point

# 5. Atualização da animação
def update_frame(frame, x, y, line, point):
    line.set_data(x[:frame], y[:frame])
    point.set_data(x[frame], y[frame])
    return line, point

# 6. Criação do objeto de animação
def create_animation(fig, ax, x, y, interval_ms):
    line, = ax.plot([], [], lw=2, color='blue')
    point, = ax.plot([], [], 'ro', markersize=6)

    ani = FuncAnimation(
        fig,
        func=lambda frame: update_frame(frame, x, y, line, point),
        init_func=lambda: init_animation(line, point),
        frames=len(x),
        interval=interval_ms,
        blit=True
    )
    return ani

# 7. Função final: montar tudo, exibir e salvar
def animate_tangent(
    a=0, b=1, c=1, d=0,
    x_min=-np.pi/2 + 0.15,
    x_max=np.pi/2 - 0.15,
    n_points=900,
    y_min=-6,
    y_max=6,
    interval_ms=20,
    duration_sec=6,         
    save_mp4=False,
    save_gif=False,
    filename_prefix="tangent"
):
    # localizar assíntotas
    ks = np.arange(-3, 4, 1)
    asymptotes = [(np.pi/2 + k*np.pi - d)/c for k in ks]
    asymptotes = [x for x in asymptotes if x_min - 1 < x < x_max + 1]

    # gerar dados
    x, y = compute_data(a, b, c, d, x_min, x_max, n_points)

    # figura
    fig, ax, asymp_lines = config_axes(x_min, x_max, y_min, y_max, asymptotes)

    # animação
    ani = create_animation(fig, ax, x, y, interval_ms)

    # garante duração idêntica no Jupyter e no MP4/GIF
    fps = n_points / duration_sec   

    # nomes dos arquivos
    base_name = f"{filename_prefix}_a{a}_b{b}_c{c}_d{d}"

    # SALVAR MP4 (via FFmpeg)
    if save_mp4:
        ani.save(f"{base_name}.mp4", writer="ffmpeg", fps=fps)
        print(f"[OK] Arquivo salvo: {base_name}.mp4 ({duration_sec}s)")

    # SALVAR GIF (PillowWriter)
    if save_gif:
        ani.save(f"{base_name}.gif", writer=PillowWriter(fps=fps))
        print(f"[OK] Arquivo salvo: {base_name}.gif ({duration_sec}s)")

    plt.close(fig)
    return HTML(ani.to_html5_video())
```

```python
ani = animate_tangent(
    a=0, b=1, c=1, d=0,
    x_min=-2*np.pi, 
    x_max=2*np.pi,
    y_min=-10,
    y_max=10,
    n_points=600,
    duration_sec=9,
    save_mp4=True,
    save_gif=False
)
```
