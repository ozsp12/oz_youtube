```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Define o comprimento do lado do quadrado
a = 1

# Cria a figura e os eixos para o gráfico
fig, ax = plt.subplots(figsize=(6, 6))

# Coordenadas dos vértices do quadrado ABCD
square_x = [0, a, a, 0, 0]
square_y = [0, 0, a, a, 0]
ax.plot(square_x, square_y, 'k')  # Desenha o quadrado em preto

# Desenha a diagonal AC do quadrado
ax.plot([0, a], [0, a], 'k', linewidth=2)

# Arco centrado no ponto D = (a, 0) — segundo quadrante de um círculo
theta_D = np.linspace(0, np.pi/2, 100)  # Ângulos de 0 a 90 graus
arc_D_x = a - np.cos(theta_D)  # Coordenadas x do arco
arc_D_y = np.sin(theta_D)      # Coordenadas y do arco
ax.plot(arc_D_x, arc_D_y, 'k')

# Arco centrado no ponto B = (0, a) — quarto quadrante de um círculo
theta_B = np.linspace(0, np.pi/2, 100)
arc_B_x = np.cos(theta_B)
arc_B_y = a - np.sin(theta_B)
ax.plot(arc_B_x, arc_B_y, 'k')

# Constrói o caminho ("folha") delimitado pelos dois arcos e a diagonal
leaf_x = np.concatenate((arc_B_x, [a], arc_D_x[::-1], [0]))
leaf_y = np.concatenate((arc_B_y, [a], arc_D_y[::-1], [0]))

# Cria um polígono preenchido com hachuras para representar a área sombreada
leaf_patch = patches.Polygon(
    xy=list(zip(leaf_x, leaf_y)),
    closed=True,
    hatch='///',          # Estilo de hachura
    edgecolor='black',
    facecolor='none',     # Sem preenchimento sólido
    linewidth=0.0
)
ax.add_patch(leaf_patch)

# Adiciona rótulos aos pontos do quadrado e pontos intermediários E e F
offset = 0.03
ax.text(0 - offset, 0 - offset, 'A', fontsize=14, color='green')
ax.text(0 - offset, a + offset, 'B', fontsize=14, color='green')
ax.text(a + offset, a + offset, 'C', fontsize=14, color='green')
ax.text(a + offset, 0 - offset, 'D', fontsize=14, color='green')
ax.text(0.25*a, 0.75*a, 'E', fontsize=14, color='green')
ax.text(0.75*a, 0.25*a, 'F', fontsize=14, color='green')

# Adiciona rótulos para os lados do quadrado
ax.text(0.5*a, a + 0.05, 'a', fontsize=14, color='green')
ax.text(0.5*a, -0.1, 'a', fontsize=14, color='green')
ax.text(-0.1, 0.5*a, 'a', fontsize=14, color='green', rotation=0)
ax.text(a + 0.05, 0.5*a, 'a', fontsize=14, color='green', rotation=0)

# Formatação final do gráfico
ax.set_aspect('equal')  # Mantém a proporção 1:1
ax.axis('off')          # Remove os eixos

# Exibe a figura
plt.show()
```
